## Etapa 4. Plan y proceso de limpieza

Se sigue el enfoque de la Clase 5 y la Clase 6: nunca se modifica `df`, todo el trabajo de limpieza se realiza sobre una copia. Cada paso revisa un aspecto especifico y documenta si encontro un problema o si la columna ya estaba correcta; no se inventan problemas que el diagnostico de la Etapa 3 no evidencio.

```python
df_limpio = df.copy()
```

### Paso 1. Normalizacion de nombres de columnas

```python
columnas_originales = list(df_limpio.columns)
df_limpio.columns = df_limpio.columns.str.strip().str.lower()
print("Columnas antes :", columnas_originales)
print("Columnas despues:", list(df_limpio.columns))
print("Se modificaron nombres de columna:", columnas_originales != list(df_limpio.columns))
```

Los nombres de columna ya estaban en minuscula, sin espacios y con guion bajo. No se requirio ningun cambio.

### Paso 2. Eliminacion de duplicados exactos

```python
filas_antes = len(df_limpio)
df_limpio = df_limpio.drop_duplicates()
filas_despues = len(df_limpio)
print("Filas antes:", filas_antes)
print("Filas despues:", filas_despues)
print("Filas eliminadas por duplicado:", filas_antes - filas_despues)
```

No se encontraron filas duplicadas para eliminar, tal como ya se habia observado en la Etapa 3.

### Paso 3. Limpieza de espacios en variables de texto

```python
columnas_texto = df_limpio.select_dtypes(include="object").columns
cambios_por_columna = {}
for columna in columnas_texto:
    valores_antes = df_limpio[columna].copy()
    df_limpio[columna] = df_limpio[columna].str.strip()
    cambios_por_columna[columna] = (valores_antes != df_limpio[columna]).sum()
print("Valores modificados por columna al quitar espacios:")
print(cambios_por_columna)
```

Ninguna columna de texto tenia espacios adicionales al inicio o al final; la operacion no modifico ningun valor.

### Paso 4. Homogeneizacion de categorias equivalentes

```python
for columna in columnas_texto:
    print(f"--- {columna} ---")
    print(sorted(df_limpio[columna].unique()))
    print()

# Verificacion cuantitativa: no basta con mirar la lista de categorias a simple
# vista. Se compara el numero de categorias unicas contra el numero de categorias
# unicas tras normalizar (quitar espacios y pasar a minusculas). Si coinciden,
# queda demostrado -- y no solo afirmado -- que no existen categorias equivalentes
# escritas de forma distinta (por ejemplo, "Femenino" / "femenino" / " femenino ").
print("Verificacion de categorias equivalentes (normalizado vs. original):")
for columna in columnas_texto:
    unicos_originales = df_limpio[columna].nunique()
    unicos_normalizados = df_limpio[columna].str.strip().str.lower().nunique()
    print(f"{columna}: {unicos_originales} categorias originales, "
          f"{unicos_normalizados} tras normalizar -> "
          f"{'sin duplicados ocultos' if unicos_originales == unicos_normalizados else 'POSIBLES DUPLICADOS'}")
```

Al revisar las categorias unicas de cada variable de texto no se encontraron categorias equivalentes escritas de forma distinta (por ejemplo, mismas categorias con mayusculas diferentes, con espacios adicionales, o abreviadas). Esto no se afirma unicamente a partir de la inspeccion visual de `unique()`: la verificacion cuantitativa muestra que el numero de categorias no cambia al normalizar (quitar espacios y pasar a minusculas), lo que confirma de forma objetiva que no hay categorias equivalentes ocultas en `género`, `uso_plataforma`, `nivel_interacción_social` ni `riesgo_depresión`. Por lo tanto, no se aplico ningun diccionario de reemplazo en este paso.

### Paso 5. Identificacion de valores imposibles

```python
# Que no haya valores nulos ni filas duplicadas (Etapa 3) no significa que los datos
# esten "limpios": un valor puede estar presente, no estar duplicado, y aun asi ser
# fisicamente o logicamente imposible (por ejemplo, mas de 24 horas de sueño en un dia,
# o una edad fuera de la poblacion de estudio). Por eso se hace una validacion de
# plausibilidad explicita sobre cada variable numerica, en lugar de asumir que "sin
# nulos y sin duplicados" ya implica calidad de los datos.
#
# Las reglas se definen solo cuando estan justificadas por el significado de la
# variable o por una restriccion fisica evidente (no se inventan escalas):
# - edad: debe estar en el rango 13-19 de la poblacion de estudio.
# - horas_sueño, horas_diarias_redes_sociales, tiempo_pantalla_antes_dormir: no pueden
#   ser negativas ni superar las 24 horas de un dia (un valor de, por ejemplo, 65 horas
#   de sueño -- como el caso mencionado en la retroalimentacion de la profesora -- seria
#   fisicamente imposible y se marcaria como faltante).
# - actividad_física, rendimiento_académico: no pueden ser negativos (no se conoce un
#   limite superior documentado, por lo que no se inventa uno).
# - nivel_estrés, nivel_ansiedad: deben respetar la escala observada/documentada de 1 a 10.
imposibles = {
    "edad fuera de 13-19": ((df_limpio["edad"] < 13) | (df_limpio["edad"] > 19)).sum(),
    "horas_sueño negativas o > 24h": ((df_limpio["horas_sueño"] < 0) | (df_limpio["horas_sueño"] > 24)).sum(),
    "horas_diarias_redes_sociales negativas o > 24h": ((df_limpio["horas_diarias_redes_sociales"] < 0) | (df_limpio["horas_diarias_redes_sociales"] > 24)).sum(),
    "tiempo_pantalla_antes_dormir negativo o > 24h": ((df_limpio["tiempo_pantalla_antes_dormir"] < 0) | (df_limpio["tiempo_pantalla_antes_dormir"] > 24)).sum(),
    "actividad_física negativa": (df_limpio["actividad_física"] < 0).sum(),
    "rendimiento_académico negativo": (df_limpio["rendimiento_académico"] < 0).sum(),
    "nivel_estrés fuera de 1-10": ((df_limpio["nivel_estrés"] < 1) | (df_limpio["nivel_estrés"] > 10)).sum(),
    "nivel_ansiedad fuera de 1-10": ((df_limpio["nivel_ansiedad"] < 1) | (df_limpio["nivel_ansiedad"] > 10)).sum(),
}
for descripcion, cantidad in imposibles.items():
    print(f"{descripcion}: {cantidad}")

print()
print("Verificacion puntual del caso mencionado por la profesora (65 horas de sueño):")
print("Maximo real de horas_sueño en el dataset:", df_limpio["horas_sueño"].max())
print("Minimo real de horas_sueño en el dataset:", df_limpio["horas_sueño"].min())
```

No se encontraron valores logicamente imposibles en ninguna de las ocho reglas de plausibilidad aplicadas: la edad se mantiene entre 13 y 19 anos, ninguna variable de horas es negativa ni supera las 24 horas de un dia, `actividad_física` y `rendimiento_académico` no tienen valores negativos, y `nivel_estrés`/`nivel_ansiedad` se mantienen en su escala de 1 a 10.

En particular, respecto al caso mencionado por la profesora en la primera presentacion (un valor como 65 horas de sueño, que no puede representar horas de sueño por dia): se comprobo mediante codigo que este valor **no existe** en `dataset_salud_mental_Alfons.csv`. El maximo real de `horas_sueño` en los datos es 9.0 horas y el minimo es 4.0 horas, ambos fisicamente plausibles. No se inventa ni se simula este error para el proyecto: la conclusion de que no hay valores imposibles se basa exclusivamente en la ejecucion del codigo anterior sobre el CSV real. Por lo tanto, no fue necesario convertir ningun valor a `np.nan` en este paso.

## Etapa 6. Preparacion de datos para el analisis

Para el analisis de correlacion y de visualizacion (Etapas 7 y 8) no se requiere ninguna transformacion adicional: las variables numericas se usan en su escala original. La unica preparacion necesaria es la creacion de tres variables categoricas derivadas -- grupos de sueno, de uso de redes sociales y de actividad fisica -- que se usaran unicamente en la Etapa 9 para comparar el rendimiento academico promedio entre grupos, siguiendo el mismo enfoque de `pd.cut` usado en la Caja de Herramientas de la clase.

```python
df_limpio["grupo_sueño"] = pd.cut(
    df_limpio["horas_sueño"], bins=[4, 6, 7, 9], include_lowest=True,
    labels=["Bajo (<6h)", "Medio (6-7h)", "Alto (>7h)"]
)

df_limpio["grupo_redes"] = pd.cut(
    df_limpio["horas_diarias_redes_sociales"], bins=[1, 3, 6, 8], include_lowest=True,
    labels=["Bajo (<=3h)", "Medio (3-6h)", "Alto (>6h)"]
)

df_limpio["grupo_actividad"] = pd.cut(
    df_limpio["actividad_física"], bins=[0, 0.7, 1.3, 2], include_lowest=True,
    labels=["Bajo", "Medio", "Alto"]
)

print(df_limpio[["horas_sueño","grupo_sueño","horas_diarias_redes_sociales","grupo_redes","actividad_física","grupo_actividad"]].head())
```

```python
print("Conteo por grupo_sueño:")
print(df_limpio["grupo_sueño"].value_counts().sort_index())
print()
print("Conteo por grupo_redes:")
print(df_limpio["grupo_redes"].value_counts().sort_index())
print()
print("Conteo por grupo_actividad:")
print(df_limpio["grupo_actividad"].value_counts().sort_index())

# Verificacion explicita de que ninguna agrupacion pierde registros: se comprueba que
# la suma de los conteos por grupo sea igual al total de filas, y que no haya valores
# faltantes generados por los limites de los intervalos de pd.cut.
print()
print("Valores faltantes por variable de grupo (deben ser 0):")
print("grupo_sueño:", df_limpio["grupo_sueño"].isna().sum())
print("grupo_redes:", df_limpio["grupo_redes"].isna().sum())
print("grupo_actividad:", df_limpio["grupo_actividad"].isna().sum())
print()
print("Filas totales:", len(df_limpio))
print("Suma grupo_sueño:", df_limpio["grupo_sueño"].value_counts().sum())
print("Suma grupo_redes:", df_limpio["grupo_redes"].value_counts().sum())
print("Suma grupo_actividad:", df_limpio["grupo_actividad"].value_counts().sum())
```

**Correccion aplicada:** en una version anterior de este notebook, los limites inferiores de `grupo_sueño` (bins=[4, 6, 7, 9]) y `grupo_redes` (bins=[1, 3, 6, 8]) no incluian el valor minimo observado de cada variable, porque `pd.cut` construye intervalos abiertos por la izquierda de forma predeterminada. Como el minimo real de `horas_sueño` es exactamente 4.0 y el minimo real de `horas_diarias_redes_sociales` es exactamente 1.0, esos registros quedaban fuera de cualquier intervalo y se convertian en valores faltantes de forma silenciosa (41 registros en `grupo_sueño` y 12 en `grupo_redes`). `grupo_actividad` no presentaba este problema porque ya se habia construido con `include_lowest=True`. La correccion consistio en aplicar ese mismo parametro, ya usado en `grupo_actividad`, a las otras dos agrupaciones. La verificacion anterior confirma que las tres variables de grupo tienen cero valores faltantes y que la suma de los conteos por categoria coincide con el total de 2500 filas, por lo que ninguna agrupacion pierde registros y las tres son utilizables sin sesgo en la comparacion de la Etapa 9.

## Etapa 7. Analisis exploratorio de datos (EDA)

Se analiza la distribucion de cada variable principal del proyecto. Cada grafico incluye titulo, nombres de ejes e interpretacion, y responde una pregunta especifica relacionada con el objetivo del proyecto.

### Deteccion de valores atipicos en las variables principales

Pregunta: hay valores atipicos en las variables de habito o en el rendimiento academico que deban tenerse en cuenta antes del analisis de relaciones?

```python
variables_principales = ["horas_sueño", "horas_diarias_redes_sociales", "actividad_física", "rendimiento_académico"]

fig, axes = plt.subplots(1, 4, figsize=(16, 4))
for ax, variable in zip(axes, variables_principales):
    sns.boxplot(data=df_limpio, y=variable, ax=ax)
    ax.set_title(variable)
plt.tight_layout()
plt.show()
```

```python
def contar_atipicos_iqr(serie):
    q1, q3 = serie.quantile([0.25, 0.75])
    iqr = q3 - q1
    limite_inferior = q1 - 1.5 * iqr
    limite_superior = q3 + 1.5 * iqr
    return ((serie < limite_inferior) | (serie > limite_superior)).sum()

for variable in variables_principales:
    print(f"{variable}: {contar_atipicos_iqr(df_limpio[variable])} valores atipicos (regla IQR)")
```

Los boxplots y el conteo de atipicos por la regla del rango intercuartilico (1.5 veces el IQR) no muestran valores atipicos relevantes en ninguna de las cuatro variables principales. ## Etapa 10. Modelo

**Seleccion del modelo.** Ninguno de los cuadernos de clase proporcionados incluye una metodologia de modelado predictivo: no aparece `scikit-learn`, ni division en conjuntos de entrenamiento y prueba, ni regresion multiple. La unica herramienta relacionada con relaciones entre variables presente en los materiales es `scipy.stats`, ya usada en la Clase 9 para pruebas de normalidad, y el calculo de correlacion de Pearson usado en la Etapa 8.

Siguiendo la instruccion de no resolver con una metodologia avanzada lo que los cuadernos no explican, se elige el modelo mas simple y justificable con las herramientas ya usadas en clase: una **regresion lineal simple** calculada con `scipy.stats.linregress`, que ajusta una recta entre una unica variable predictora y la variable de resultado.

**Variable predictora.** De las tres variables de habito, `horas_diarias_redes_sociales` es la que presento la correlacion mas fuerte con `rendimiento_académico` en la Etapa 8 (r = -0.863), por lo que se usa como predictor del modelo. No se eligio esta variable de antemano: la seleccion se baso en el resultado de la Etapa 8.

```python
x = df_limpio["horas_diarias_redes_sociales"]
y = df_limpio["rendimiento_académico"]

modelo = stats.linregress(x, y)
print(f"Pendiente (slope): {modelo.slope:.4f}")
print(f"Intercepto: {modelo.intercept:.4f}")
print(f"Coeficiente de correlación (r): {modelo.rvalue:.4f}")
print(f"R al cuadrado: {modelo.rvalue**2:.4f}")
print(f"p-valor: {modelo.pvalue:.4g}")
print(f"Error estándar de la pendiente: {modelo.stderr:.4f}")
```

El modelo ajustado es: `rendimiento_académico = intercepto + pendiente * horas_diarias_redes_sociales`. La pendiente negativa confirma la direccion de la relacion observada en la Etapa 8: por cada hora adicional de uso diario de redes sociales, el rendimiento academico promedio predicho por el modelo disminuye en el valor de la pendiente.

```python
predicciones = modelo.intercept + modelo.slope * x

fig, ax = plt.subplots(figsize=(7, 5))
ax.scatter(x, y, alpha=0.3, label="datos observados")
orden = x.sort_values().index
ax.plot(x[orden], predicciones[orden], color="black", linewidth=2, label="recta ajustada")
ax.set_title("Modelo de regresión lineal simple: rendimiento académico vs. horas en redes sociales")
ax.set_xlabel("horas diarias en redes sociales")
ax.set_ylabel("rendimiento académico")
ax.legend()
plt.show()
```

La recta ajustada sigue visualmente la tendencia de los datos observados, lo cual es consistente con el valor de r reportado.

## Etapa 11. Evaluación e interpretación del modelo

Se evalua el modelo con dos metricas simples, calculadas directamente a partir de las predicciones: el coeficiente de determinacion (R^2), que indica que proporcion de la variacion del rendimiento academico es explicada linealmente por las horas en redes sociales, y el error absoluto medio (MAE), que indica en promedio que tan lejos estan las predicciones del valor real, en la misma escala que `rendimiento_académico`.

**Advertencia metodologica:** el modelo se ajusta y se evalua sobre el mismo conjunto de datos (no se separo una muestra de entrenamiento y otra de prueba, siguiendo el alcance de los materiales de la Clase 9 y del Taller de Buenas Graficas, que no incluyen esta tecnica). Por lo tanto, el R^2 reportado a continuacion debe interpretarse unicamente como una medida de que tan bien se ajusta la recta a los datos observados (ajuste dentro de muestra), y no como una medida de desempeño predictivo del modelo sobre datos nuevos o no vistos.

```python
residuos = y - predicciones
mae = residuos.abs().mean()

print(f"R al cuadrado: {modelo.rvalue**2:.4f}")
print(f"Error absoluto medio (MAE): {mae:.4f}")
```

```python
fig, ax = plt.subplots(figsize=(7, 4.5))
sns.histplot(residuos, kde=True, ax=ax)
ax.set_title("Distribución de los residuos del modelo")
ax.set_xlabel("residuo (valor real - valor predicho)")
ax.set_ylabel("cantidad de adolescentes")
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(7, 4.5))
ax.scatter(predicciones, residuos, alpha=0.3)
ax.axhline(0, color="black", linewidth=1)
ax.set_title("Residuos vs. valores ajustados")
ax.set_xlabel("valor ajustado (predicho) de rendimiento_académico")
ax.set_ylabel("residuo (valor real - valor predicho)")
plt.show()
```

El grafico de residuos contra valores ajustados se revisa para comprobar visualmente si existe algun patron evidente (por ejemplo, forma de embudo o de curva) que sugiera que el modelo lineal no es adecuado. En esta revision exploratoria no se observa un patron evidente de curvatura ni de aumento sistematico de la dispersion a lo largo del eje horizontal; los residuos se distribuyen de forma aproximadamente pareja alrededor de cero. Esto no demuestra que se cumplan de forma perfecta los supuestos de un modelo lineal, pero tampoco muestra una senal clara de que el ajuste lineal simple sea inadecuado para este analisis exploratorio.

El modelo explica una proporcion importante de la variacion del rendimiento academico a partir de una unica variable (las horas diarias en redes sociales), lo cual es consistente con la correlacion fuerte encontrada en la Etapa 8. El error absoluto medio indica el margen tipico de desviacion entre el rendimiento predicho y el rendimiento real reportado en el dataset. Los residuos se distribuyen de forma aproximadamente simetrica alrededor de cero, y el grafico de residuos vs. valores ajustados no muestra un patron evidente, sin un problema grave de ajuste para este modelo simple.

**El R^2 reportado debe leerse como una medida de ajuste dentro de la muestra utilizada para entrenar el modelo, no como una medida de desempeño predictivo fuera de muestra**: al no existir una separacion entre datos de entrenamiento y de prueba, no es posible afirmar que el modelo prediria con esa misma precision el rendimiento academico de adolescentes que no forman parte de este dataset.

Este resultado debe interpretarse como una asociacion lineal capturada por el modelo, no como una relacion causal: el modelo no permite afirmar que reducir las horas en redes sociales vaya a producir, para un adolescente en particular, el aumento de rendimiento que sugiere la pendiente.
